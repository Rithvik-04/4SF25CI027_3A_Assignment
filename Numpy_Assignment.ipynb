* 1. Array Creation and Manipulation
Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.

```python
import numpy as np

arr = np.random.randint(1, 21, (5, 5))

print("Original Array:")
print(arr)

arr[:, 2] = 1

print("After replacing third column:")
print(arr)
```

* 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

np.fill_diagonal(arr, 0)

print("After replacing diagonal:")
print(arr)
```

* 3. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

```python
import numpy as np

arr = np.random.randint(1, 20, (5, 5))

print("Array:")
print(arr)

border = np.concatenate([
    arr[0, :],
    arr[-1, :],
    arr[1:-1, 0],
    arr[1:-1, -1]
])

print("Border elements:")
print(border)
```

* 4. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print(arr)

row_sum = np.sum(arr, axis=1)
column_sum = np.sum(arr, axis=0)

print("Row-wise sum:", row_sum)
print("Column-wise sum:", column_sum)
```

* 5. Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).


```python
import numpy as np

arr = np.arange(1, 10).reshape(3, 3)

print("Original:")
print(arr)

arr1 = arr.reshape(1, 9)
print("Shape (1,9):")
print(arr1)

arr2 = arr.reshape(9, 1)
print("Shape (9,1):")
print(arr2)
```

* 6. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
import numpy as np

data = np.array([
    ("Raj", 21, 65.5),
    ("Ravi", 19, 60.2),
    ("Kiran", 23, 70.1),
    ("Manish", 20, 68.4)
], dtype=[
    ("name", "U20"),
    ("age", "i4"),
    ("weight", "f4")
])

print("Original:")
print(data)

sorted_data = np.sort(data, order="age")

print("Sorted by age:")
print(sorted_data)
```
